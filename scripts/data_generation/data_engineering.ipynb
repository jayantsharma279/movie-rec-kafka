````python
import os
from datetime import datetime
from json import dumps, loads
from time import sleep
from random import randint
from kafka import KafkaConsumer
from typing import Dict, Any
import re

# import pandas as pd (optional to run the conversion script)
````

````python
consumer = KafkaConsumer(
    'movielog17',
    bootstrap_servers=['localhost:9092'],
    auto_offset_reset='latest',
    enable_auto_commit=True,
    auto_commit_interval_ms=1000
)

print('Reading Kafka Broker...')

with open("movie_ratings_clean.csv", "a", encoding="utf-8") as f_ratings, \
     open("movie_watch_clean.csv", "a", encoding="utf-8") as f_watch:

    f_ratings.write("userid,movie_name,year,rating\n") #whenever /rate is found
    f_watch.write("userid,movie_name,year,minute\n") #90% of the data (streaming info)

    for msg in consumer:
        line = msg.value.decode("utf-8", errors="replace").strip()
        parts = line.split(",", 2)

        if len(parts) < 3:
            continue

        userid = parts[1].strip()
        rest = parts[2].strip()

        # ---------------------------
        # Case 1: Ratings (/rate/)
        # ---------------------------
        if rest.startswith("GET /rate/"):
            raw = rest.replace("GET /rate/", "")

            if "=" in raw:
                movie_with_year, rating = raw.split("=", 1)

                match = re.match(r"(.+)(\d{4})$", movie_with_year)
                if match:
                    movie_raw = match.group(1).rstrip("+")
                    year = match.group(2)

                    movie_name = movie_raw.replace("+", " ").strip()

                    final_line = f"{userid},{movie_name},{year},{rating.strip()}"
                    print("[RATING]", final_line)
                    f_ratings.write(final_line + "\n")

        # ---------------------------
        # Case 2: Watch events (/data/m/)
        # ---------------------------
        elif rest.startswith("GET /data/m/"):
            raw = rest.replace("GET /data/m/", "") 

            if "/" in raw:
                movie_with_year, minute_part = raw.split("/", 1)

                minute = minute_part.replace(".mpg", "").strip()
                match = re.match(r"(.+)(\d{4})$", movie_with_year)
                if match:
                    movie_raw = match.group(1).rstrip("+")
                    year = match.group(2)

                    movie_name = movie_raw.replace("+", " ").strip()

                    final_line = f"{userid},{movie_name},{year},{minute}"
                    print("[WATCH]", final_line)
                    f_watch.write(final_line + "\n")

````
